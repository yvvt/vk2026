3. Загрузка и подготовка данных

```python
print("Загрузка датасетов...")

# Загружаем обучающий датасет (подвыборка)
train_dataset_full = load_dataset("deepvk/LLaVA-Instruct-ru", split="train")
print(f"Всего записей в llava-instruction-ru: {len(train_dataset_full)}")

# Берем 2000 записей для быстрого обучения
NUM_TRAIN_SAMPLES = 50 #2000
train_dataset = train_dataset_full.shuffle(seed=42).select(range(NUM_TRAIN_SAMPLES))
print(f"Используем для обучения: {len(train_dataset)} записей")

# Загружаем бенчмарк для оценки (полный)
eval_dataset = load_dataset("deepvk/mmbench-ru", split="dev")
print(f"Записей для оценки (mmbench-ru): {len(eval_dataset)}")
```

4. Предобработка данных для обучения

```python
"""4. Предобработка данных для обучения"""

print("Проверка типа изображения в датасете:")
print(f"Тип example['image']: {type(train_dataset[0]['image'])}")

# Декодируем строковые пути в PIL Image, если это еще не сделано
if isinstance(train_dataset[0]['image'], str):
    print("Декодирование изображений из строк в PIL Image...")
    from datasets import Image
    train_dataset = train_dataset.cast_column("image", Image())
    print(f"После декодирования тип: {type(train_dataset[0]['image'])}")

def format_conversations(example):
    """
    Форматирует ТОЛЬКО текст.
    Мы НЕ вкладываем PIL Image в словарь messages, чтобы избежать
    ошибки сериализации JSON/Arrow в библиотеке datasets.
    """
    messages = []
    for turn in example['conversations']:
        role = 'user' if turn['from'] == 'human' else 'assistant'
        # Убираем тег <image>, так как мы добавим изображение программно в коллейторе
        text = turn['value'].replace('<image>', '').strip()
        messages.append({"role": role, "content": text})

    # Возвращаем только messages. Колонка 'image' останется в датасете,
    # так как мы не указали её в remove_columns.
    return {"messages": messages}

print("Форматирование датасета (без вложенных изображений)...")
train_dataset_formatted = train_dataset.map(
    format_conversations,
    remove_columns=['conversations', 'type', 'id'], # 'image' намеренно НЕ удаляем
    num_proc=2
)

print("Форматирование завершено успешно.")
print(f"Колонки датасета: {train_dataset_formatted.column_names}")
print(f"Тип image: {type(train_dataset_formatted[0]['image'])}")
```

5. Загрузка модели с QLoRA

```python
print("Загрузка модели Qwen2-VL-2B с 4-битным квантованием...")

# Конфигурация квантования
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

# Загружаем модель
model = Qwen2VLForConditionalGeneration.from_pretrained(
    "Qwen/Qwen2-VL-2B-Instruct",
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    attn_implementation="sdpa" if torch.cuda.is_available() else "eager"
)

# Загружаем процессор
processor = AutoProcessor.from_pretrained("Qwen/Qwen2-VL-2B-Instruct")
processor.model_max_length = 1024

# Подготавливаем модель для k-bit обучения
model = prepare_model_for_kbit_training(model)

print(f"Модель загружена. Параметры: {model.get_input_embeddings().num_embeddings}")
```

6. Настройка LoRA

```python
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]
)

# Применяем LoRA к модели
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
```

7. Настройка параметров обучения

```python
training_args = TrainingArguments(
    output_dir="./qwen2vl-2b-ru-lora",
    per_device_train_batch_size=2, #2
    gradient_accumulation_steps=4, #4
    learning_rate=2e-4,
    num_train_epochs=1,
    fp16=False,
    bf16=True, #True
    logging_steps=10,
    save_strategy="epoch",
    optim="paged_adamw_32bit", #"paged_adamw_32bit"
    gradient_checkpointing=True, #True
    max_grad_norm=0.3,
    warmup_ratio=0.03,
    lr_scheduler_type="cosine",
    report_to="none",
    save_total_limit=1,
    dataloader_num_workers=2, #2
    remove_unused_columns=False,
    # max_seq_length=1024,
)

print(f"Параметры обучения настроены")
print(f"Эпох: {training_args.num_train_epochs}")
print(f"Batch size: {training_args.per_device_train_batch_size}")
print(f"Gradient accumulation: {training_args.gradient_accumulation_steps}")
```

8. Создание Trainer и обучение

```python
"""8. Создание Trainer и обучение"""

def data_collator(features):
    """
    Кастомный коллейтор, который динамически собирает мультимодальный батч.
    """
    batch_images = []
    batch_messages = []

    for f in features:
        # 1. Берем изображение из корневой колонки (оно там есть благодаря Пункту 4)
        img = f["image"]
        batch_images.append(img)

        # 2. Динамически внедряем изображение в первое сообщение пользователя
        msgs = f["messages"]
        for msg in msgs:
            if msg["role"] == "user":
                original_text = msg["content"]
                # Формируем структуру, которую ожидает Qwen2-VL processor
                msg["content"] = [
                    {"type": "image", "image": img},
                    {"type": "text", "text": original_text}
                ]
                break # Внедряем только в первый user-turn (стандарт LLaVA)

        # 3. Применяем chat template для получения итогового строкового промпта
        text = processor.apply_chat_template(
            msgs,
            tokenize=False,
            add_generation_prompt=False
        )
        batch_messages.append(text)

    # 4. Токенизация и обработка изображений процессором
    batch = processor(
        text=batch_messages,
        images=batch_images,
        return_tensors="pt",
        padding=True
    )

    # 5. КРИТИЧЕСКИ ВАЖНО: Маскирование меток (Label Masking)
    labels = batch["input_ids"].clone()

    for i, feature in enumerate(features):
        # Токенизируем только промпт (первое сообщение), чтобы найти его длину
        prompt_msgs = [feature["messages"][0]] # Только первый turn (user)
        # Восстанавливаем структуру с изображением для корректного расчета длины промпта
        prompt_msgs[0]["content"] = [
            {"type": "image", "image": feature["image"]},
            {"type": "text", "text": feature["messages"][0]["content"][1]["text"]}
        ]
        prompt_text = processor.apply_chat_template(
            prompt_msgs,
            tokenize=False,
            add_generation_prompt=True
        )
        prompt_len = len(processor.tokenizer(prompt_text, add_special_tokens=False).input_ids)

        # Маскируем промпт (модель не должна учиться его предсказывать)
        labels[i, :prompt_len] = -100

    # Маскируем pad-токены
    labels[batch["attention_mask"] == 0] = -100
    batch["labels"] = labels

    return batch

# Используем базовый Trainer, а не SFTTrainer, чтобы иметь полный контроль над data_collator
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset_formatted,
    data_collator=data_collator,
)
```

```python
print("Начало обучения...")
trainer.train()
```

10. Оценка на MMBench-ru

```python
def evaluate_mmbench(model, processor, dataset, max_samples=50):
    """
    Оценка на бенчмарке MMBench-ru.
    АДАПТИРОВАНО ДЛЯ CPU
    """
    if max_samples:
        dataset = dataset.select(range(max_samples))

    correct = 0
    total = 0
    skipped = 0
    model.eval()

    print(f"Начало оценки на {len(dataset)} примерах...")

    for i, example in enumerate(dataset):
        if i % 10 == 0 and i > 0:  # Уменьшил частоту логов для CPU
            print(f"Обработано: {i}/{len(dataset)}, Верно: {correct}, Пропущено: {skipped}")

        question = example["question"]
        hint = example.get("hint", None)

        # Варианты ответов
        options = {
            "A": example.get("A", ""),
            "B": example.get("B", ""),
            "C": example.get("C", ""),
            "D": example.get("D", "")
        }
        answer = example["answer"]

        # Формируем текст вариантов
        options_text = "\n".join([f"{key}. {val}" for key, val in options.items() if val])

        # Формируем промпт
        prompt = f"{question}\n"
        if hint and str(hint).strip() != "nan" and str(hint).strip() != "":
            prompt += f"Подсказка: {hint}\n"
        prompt += f"\nВарианты ответов:\n{options_text}\n\nОтветь только одной буквой (A, B, C или D)."

        # Загружаем изображение
        try:
            image = example["image"]
            if isinstance(image, str):
                # Если это путь к локальному файлу
                if os.path.exists(image):
                    image = Image.open(image).convert("RGB")
                else:
                    # Если это URL
                    image = Image.open(BytesIO(requests.get(image).content)).convert("RGB")
        except Exception as e:
            print(f"Ошибка загрузки изображения: {e}")
            skipped += 1
            continue

        # Формируем messages
        messages = [
            {
                "role": "user",
                "content": [
                    {"type": "image", "image": image},
                    {"type": "text", "text": prompt}
                ]
            }
        ]

        # Токенизация
        text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = processor(text=[text], images=[image], return_tensors="pt", padding=True)

        # ДЛЯ CPU: не используем .to(model.device), оставляем на CPU
        inputs = inputs.to(model.device)  # ЗАКОММЕНТИРОВАНО ДЛЯ CPU

        # Генерация
        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=5,
                do_sample=False,
                temperature=0.1,
                pad_token_id=processor.tokenizer.eos_token_id
            )

        # Декодируем
        generated_text = processor.decode(outputs[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True).strip()

        # Извлекаем букву
        generated_answer = None
        generated_text_upper = generated_text.upper()
        for letter in ["A", "B", "C", "D"]:
            if letter in generated_text_upper:
                generated_answer = letter
                break

        # Сравниваем
        true_answer = str(answer).strip().upper().replace(".", "")
        if generated_answer == true_answer:
            correct += 1
        total += 1

        if max_samples and total >= max_samples:
            break

    accuracy = correct / total if total > 0 else 0
    return accuracy, correct, total
```

11. Запуск оценки

```python
print("Загрузка базовой модели для оценки...")

# ДЛЯ CPU: убираем device_map и используем float32 вместо bfloat16
model_eval = Qwen2VLForConditionalGeneration.from_pretrained(
    "Qwen/Qwen2-VL-2B-Instruct",
    quantization_config=bnb_config,  # ЗАКОММЕНТИРОВАНО - не нужно для оценки LoRA
    torch_dtype=torch.bfloat16,  # float32 для CPU
    device_map='auto'  # Явно указываем None для CPU
)

# Загружаем LoRA адаптер
from peft import PeftModel
print(f"Загрузка LoRA адаптера из {output_dir}...")
model_eval = PeftModel.from_pretrained(model_eval, output_dir)

# НЕ делаем merge_and_unload() на CPU - это съест всю память
model_eval = model_eval.merge_and_unload()

print("Запуск оценки на MMBench-ru (это займет время на CPU)...")
print("Рекомендую начать с max_samples=50 для теста")

# ДЛЯ БЫСТРОГО ТЕСТА: поставьте max_samples=50
# ДЛЯ ПОЛНОЙ ОЦЕНКИ: уберите max_samples или поставьте None
accuracy, correct, total = evaluate_mmbench(model_eval, processor, eval_dataset, max_samples=100)

print(f"\n{'='*50}")
print(f"РЕЗУЛЬТАТЫ ОЦЕНКИ НА MMBENCH-RU")
print(f"{'='*50}")
print(f"Обработано примеров: {total}")
print(f"Правильных ответов: {correct}")
print(f"Итоговая Accuracy: {accuracy*100:.2f}%")
print(f"{'='*50}")

# Сохраняем результаты
results = {
    "accuracy_percent": round(accuracy*100, 2),
    "correct_answers": correct,
    "total_evaluated": total,
    "base_model": "Qwen/Qwen2-VL-2B-Instruct",
    "method": "LoRA (CPU training)",
    "train_samples_used": NUM_TRAIN_SAMPLES,
    "training_device": "CPU"
}

with open("evaluation_results.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"Результаты сохранены в evaluation_results.json")
```
